```python
import os
import csv

# Open the output CSV file where the results will be saved
with open('transcription_small_tt_nowordrep.csv', mode='w', newline='', encoding='utf-8') as output_csv:
    fieldnames = ['filepath', 'result']
    writer = csv.DictWriter(output_csv, fieldnames=fieldnames)

    # Write header row
    writer.writeheader()

    # Open and read the input CSV file (filtered_for_whisper.csv)
    with open('filtered_for_whisper.csv', mode='r', encoding='utf-8') as input_csv:
        csv_reader = csv.DictReader(input_csv)
        
        # Iterate over each row in the input CSV
        for row in csv_reader:
            # Get the values of Show, EpId, and ClipId from the row
            show = row['Show']
            ep_id = row['EpId']
            clip_id = row['ClipId']
            
            # Construct the file path: Show/EpId/ClipId.wav
            fp = os.path.join("/home/alien/Git/DATA/ml-stuttering-events-dataset/CLIP_DIR/", show, ep_id, f"{show+"_"+ep_id+"_"+clip_id}.wav")
            
            # Check if the file exists before proceeding
            if os.path.exists(fp):
                # Transcribe the audio file
                result = transcribe(fp)
                transcription_text = result
                
                # Write the result to the output CSV file
                writer.writerow({'filepath': fp, 'result': transcription_text})
                print(fp, transcription_text)
            else:
                print(f"File {fp} does not exist.")
```

```python
import csv
import string

# Define the file path
input_file_path = 'transcription_small_tt_nowordrep.csv'  # Replace with your file path
output_file_path = 'transcription_small_tt_nowordrep_formatted.csv'  # This will be the output file path

# Create a translation table to remove punctuation
remove_punctuation = str.maketrans('', '', string.punctuation)

# Read the input CSV file
with open(input_file_path, 'r', newline='', encoding='utf-8') as infile:
    csvreader = csv.DictReader(infile)
    fieldnames = csvreader.fieldnames

    # Open the output CSV file for writing
    with open(output_file_path, 'w', newline='', encoding='utf-8') as outfile:
        csvwriter = csv.DictWriter(outfile, fieldnames=fieldnames)
        csvwriter.writeheader()

        # Iterate over each row in the CSV
        for row in csvreader:
            # Remove punctuation only from the 'result' column (adjust the column name if necessary)
            if 'result' in row:
                row['result'] = row['result'].translate(remove_punctuation)
            
            # Write the modified row to the output file
            csvwriter.writerow(row)

print(f"Punctuation marks removed from the 'result' column. Modified content saved to {output_file_path}")
```
