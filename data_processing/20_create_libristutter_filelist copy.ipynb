```python
import os
import csv

def find_wav_files(directory):
    wav_files = []
    for root, dirs, files in os.walk(directory):
        for file in files:
            if file.endswith(".wav"):
                full_path = os.path.join(root, file)
                wav_files.append(full_path)
    return wav_files

def write_to_csv(file_paths, output_csv):
    with open(output_csv, mode='w', newline='') as file:
        writer = csv.writer(file)
        writer.writerow(["stuttered_fp"])  # Write header
        for path in file_paths:
            writer.writerow([path])

if __name__ == "__main__":
    directory = "/home/alien/Git/DATA/LibriStutterData/LibriStutter_16kHz"
    output_csv = "./Libristutter_16hkz_fps.csv"

    wav_files = find_wav_files(directory)
    write_to_csv(wav_files, output_csv)

    print(f"Found {len(wav_files)} .wav files. File paths saved to {output_csv}.")
```

```python
import pandas as pd

# Load the CSV file
df = pd.read_csv('Libristutter_16hkz_fps.csv')

# Create a new column with the modified file paths
df['transcript_fp'] = df['stuttered_fp'].apply(lambda x: x.replace('LibriStutter_16kHz', 'LibriStutter Transcripts').replace('.wav', '.txt'))

# Save the modified DataFrame to a new CSV file
df.to_csv('Libristutter_16hkz_fps.csv', index=False)
```
