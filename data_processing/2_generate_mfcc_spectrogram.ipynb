```python
root_fp = "../../DATA/ml-stuttering-events-dataset/"
CLIPS_DIR = "/home/alien/Git/DATA/ml-stuttering-events-dataset/CLIP_DIR/"
df = pd.read_csv("/home/alien/Git/DATA/ml-stuttering-events-dataset/SEP-28k_labels.csv")
df.head()
df.shape
```

```python
df['Name'] = df[df.columns[0:3]].apply(
    lambda x: '_'.join(x.dropna().astype(str)),
    axis=1
)
df.head()
```

```python
matplotlib.use('Agg')
def gen_image(fp, save_name, label):
    y, sr = librosa.load(fp, res_type='kaiser_fast', sr=None)
    spectrogram = np.abs(librosa.stft(y))
    mfccs = librosa.feature.mfcc(S=librosa.power_to_db(spectrogram), sr=sr, n_mfcc=13)
    plt.figure(figsize=(6, 4))
    librosa.display.specshow(mfccs, sr=sr)
    plt.tight_layout()
    if label == 0:
        save_name += "_fluent"
    if label == 1:
        save_name += "_stutter"
    plt.savefig("/home/alien/Git/DATA/mfcc_images_sep28k/" + save_name + ".jpg", bbox_inches='tight', pad_inches=0)  # Save the MFCC plot as JPG
    plt.close()

# gen_image("/home/alien/Git/DATA/ml-stuttering-events-dataset/CLIP_DIR/MyStutteringLife/20/MyStutteringLife_20_4.wav", "mfcc", 1)
```

```python
matplotlib.use('Agg')
def gen_image_five(fp, save_name, label):
    y, sr = librosa.load(fp, res_type='kaiser_fast', sr=None)
    spectrogram = np.abs(librosa.stft(y))
    mfccs = librosa.feature.mfcc(S=librosa.power_to_db(spectrogram), sr=sr, n_mfcc=13)
    plt.figure(figsize=(6, 4))
    librosa.display.specshow(mfccs, sr=sr)
    plt.tight_layout()
    if label == 0:
        save_name += "_fluent"
    if label == 1:
        save_name += "_soundrep"
    if label == 2:
        save_name += "_wordrep"
    if label == 3:
        save_name += "_block"
    if label == 4:
        save_name += "_prolongation"
    if label == 5:
        save_name += "_interjection"
    plt.savefig("/home/alien/Git/DATA/mfcc_images_sep28k_allfive/" + save_name + ".jpg", bbox_inches='tight', pad_inches=0)  # Save the MFCC plot as JPG
    plt.close()

# gen_image("/home/alien/Git/DATA/ml-stuttering-events-dataset/CLIP_DIR/MyStutteringLife/20/MyStutteringLife_20_4.wav", "mfcc", 1)
```
