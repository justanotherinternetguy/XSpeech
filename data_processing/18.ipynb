```python
import os
import pandas as pd

# Read the CSV file
csv_file = '/home/alien/Git/XSpeech/data_processing/transcription_small_tt_nowordrep_formatted.csv'  # Replace with the actual path to your CSV
df = pd.read_csv(csv_file)

# Create the "transcriptions" folder if it doesn't exist
transcriptions_folder = '/home/alien/Git/DATA/Sep28kNoWordRepTranscriptions'
if not os.path.exists(transcriptions_folder):
    os.makedirs(transcriptions_folder)

# Loop through the rows in the dataframe
for index, row in df.iterrows():
    result_text = row['result']
    audio_name = row['filepath'].split('/')[-1].replace('.wav', '.txt')  # Extract file name and change extension
    transcription_path = os.path.join(transcriptions_folder, audio_name)

    # Write the result text into a new text file
    with open(transcription_path, 'w') as file:
        file.write(result_text)

    # Replace the "result" column with the path to the text file
    df.at[index, 'result'] = transcription_path

# Save the updated dataframe back to CSV
df.to_csv('updated_csv.csv', index=False)
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split

# Load the data from the CSV file
data = pd.read_csv('updated_csv.csv')

# Split the data into training and testing sets (80-20 split)
train_data, test_data = train_test_split(data, test_size=0.2, random_state=42)

# Save the train and test sets to new CSV files
train_data.to_csv('train.csv', index=False)
test_data.to_csv('test.csv', index=False)

print("Data has been split into 'train.csv' and 'test.csv'.")
```

```python
import os
from num2words import num2words
import re

# Function to convert numerals to words
def convert_numerals_to_words(text):
    # Regular expression to find numerals
    numeral_pattern = r'\b\d+\b'
    
    # Function to convert a match to words
    def replace_numeral(match):
        num = match.group()
        return num2words(int(num))  # Convert number to words
    
    # Replace all numerals in the text with their word equivalents
    return re.sub(numeral_pattern, replace_numeral, text)

# Function to process files in the given folder
def process_folder(folder_path):
    # Loop through all .txt files in the folder
    for filename in os.listdir(folder_path):
        if filename.endswith('.txt'):
            file_path = os.path.join(folder_path, filename)
            
            # Read the file content
            with open(file_path, 'r', encoding='utf-8') as file:
                file_content = file.read()
            
            # Convert numerals to words
            updated_content = convert_numerals_to_words(file_content)
            
            # Write the updated content back to the file
            with open(file_path, 'w', encoding='utf-8') as file:
                file.write(updated_content)

# Example usage
folder_path = '/home/alien/Git/DATA/Sep28kNoWordRepTranscriptions'  # Replace with the folder path
process_folder(folder_path)
```

```python
import os

def remove_spaces_from_files(folder_path):
    # Iterate through all files in the folder
    for filename in os.listdir(folder_path):
        # Check if the file is a .txt file
        if filename.endswith('.txt'):
            file_path = os.path.join(folder_path, filename)
            
            # Open the file and read its content
            with open(file_path, 'r') as file:
                lines = file.readlines()
            
            # Strip leading and trailing spaces from each line
            stripped_lines = [line.strip() for line in lines]
            
            # Write the stripped lines back to the file
            with open(file_path, 'w') as file:
                file.writelines('\n'.join(stripped_lines))

# Specify the folder path containing your .txt files
folder_path = '/home/alien/Git/DATA/Sep28kNoWordRepTranscriptions'

# Call the function to remove spaces
remove_spaces_from_files(folder_path)
```

```python
import os
import re

def find_txt_files_with_numbers(directory):
    # List to store paths of files that contain numbers
    files_with_numbers = []

    # Traverse the directory
    for root, dirs, files in os.walk(directory):
        for file in files:
            if file.endswith('.txt'):  # Check if the file is a .txt file
                file_path = os.path.join(root, file)
                # Open the file and check if it contains numbers
                with open(file_path, 'r') as f:
                    content = f.read()
                    if re.search(r'\d', content):  # Search for any digit
                        files_with_numbers.append(file_path)

    return files_with_numbers

# Example usage
directory = '/home/alien/Git/DATA/Sep28kNoWordRepTranscriptions'
files_with_numbers = find_txt_files_with_numbers(directory)

# Print the paths of the files containing numbers
for file in files_with_numbers:
    print(file)
```
