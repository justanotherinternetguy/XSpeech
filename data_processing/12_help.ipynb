```python
import pandas as pd
import os

# Step 1: Read the CSV file
input_file = './modified_file.csv'  # Replace with your input file path
df = pd.read_csv(input_file)

# Step 2: Copy the "filepath" column
df['original_filepath'] = df['filepath']

# Step 3: Generate new file paths
def generate_new_filepath(original_path):
    # Extract the relevant parts of the original path
    parts = original_path.split('/')
    # Get the filename (last part of the path)
    filename = parts[-1]
    # Construct the new path
    new_path = os.path.join('/home/alien/Git/DATA/MeloTTSAudioLibriStutter', filename)
    return new_path

# Create a new column with the generated file paths
df['new_filepath'] = df['original_filepath'].apply(generate_new_filepath)

# Step 4: Save the new DataFrame to a new CSV file
output_file = 'output.csv'  # Replace with your desired output file path
df.to_csv(output_file, index=False)

print(f"New file created: {output_file}")
```
